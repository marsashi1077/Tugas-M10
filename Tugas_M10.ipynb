```python
# nomer 3.1

def luas_persegi(sisi):
  hasil = sisi * sisi
  print(f"Luas Persegi adalah: {hasil}")
  return hasil

def rata_rata(angka1, angka2, angka3):
  rata = (angka1 + angka2 + angka3) / 3
  return rata

angka1 = luas_persegi(5)
angka2 = luas_persegi(10)
angka3 = luas_persegi(15)
print(f"rata rata dari ketiga luas: {rata_rata(angka1, angka2, angka3)}")
```
